```python
num_tata = Tata.select_dtypes(include=np.number).columns
num_tata
```

```python
# Histogram
Tata.hist(figsize=(15,10))
plt.show()
```

```python
# Boxplot
for col in num_tata:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=Tata[col])
    plt.title(col)
    plt.show()
```

```python
cat_cols = Tata.select_dtypes(include='object').columns
cat_cols
```

```python
# Countplot
for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(data=Tata, x=col)
    plt.xticks(rotation=45)
    plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(Tata.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(Tata['Income'], bins=30, kde=True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(Tata['Credit_Score'], bins=30, kde=True)
plt.show()
```
